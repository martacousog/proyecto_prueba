```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm
```

```python
# Ejercicio 1 y 2 : Edades
edades = np.array([22, 25, 27, 29, 25, 30, 35, 25, 40, 28])

# Medidas de tendencia central
media_edades =np.mean(edades)
mediana_edades = np.median(edades)
moda_edades = stats.mode(edades).mode

# Medidas de dispersión
rango_edades = np.ptp(edades)  # o max(edades) - min(edades)
varianza_edades = np.var(edades, ddof=1)  # ddof=1 para muestra
std_edades = np.std(edades, ddof=1)

print("=== EJERCICIO 1 & 2: EDADES ===")
print(f"Media: {media_edades}")
print(f"Mediana: {mediana_edades}")
print(f"Moda: {moda_edades}")
print(f"Rango: {rango_edades}")
print(f"Varianza muestral: {varianza_edades:.2f}")
print(f"Desviación típica muestral: {std_edades:.2f}\n")
```

```python
# Ejercicio 3: Cuartiles y percentiles (Ventas)
ventas = np.array([120, 180, 200, 250, 260, 290, 320, 350, 420, 500])

q1 = np.percentile(ventas, 25)
q2 = np.percentile(ventas, 50)  # Coincide con la mediana
q3 = np.percentile(ventas, 75)
p90 = np.percentile(ventas, 90)

print("=== EJERCICIO 3: VENTAS ===")
print(f"Q1 (Percentil 25): {q1}")
print(f"Q2 (Mediana / Percentil 50): {q2}")
print(f"Q3 (Percentil 75): {q3}")
print(f"Percentil 90: {p90}")
```

```python
# Configuración de estilo
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# ---------------------------------------------------------
# Ejercicio 4: Histogramas (100 Salarios con sesgo a la derecha)
# ---------------------------------------------------------
np.random.seed(42)  # Para reproducibilidad
salarios = np.random.gamma(shape=2, scale=15, size=100) * 1000 + 18000

sns.histplot(salarios, kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Ejercicio 4: Distribución de Salarios', fontsize=12)
axes[0].set_xlabel('Salario (€)')
axes[0].set_ylabel('Frecuencia')
```

```python

# Ejercicio 5: Boxplot y Outliers (Precios de Viviendas)
precios = np.random.normal(loc=250000, scale=40000, size=95)
outliers = np.array([450000, 480000, 520000, 550000, 100000])  # Outliers
precios = np.concatenate([precios, outliers])

sns.boxplot(x=precios, ax=axes[1], color='salmon')
axes[1].set_title('Ejercicio 5: Boxplot de Precios de Viviendas', fontsize=12)
axes[1].set_xlabel('Precio de Vivienda (€)')

plt.tight_layout()
plt.show()

# Cálculos de apoyo para la explicación
media_p = np.mean(precios)
mediana_p = np.median(precios)

print(f"Media de Precios: {media_p:,.2f} €")
print(f"Mediana de Precios: {mediana_p:,.2f} €")
```

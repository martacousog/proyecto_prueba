```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
# Dataset de prueba (edades)
datos_edades = [20, 25, 22, 30, 28, 25, 35, 40, 25, 28]
df = pd.DataFrame({'Edades': datos_edades})
```

```python
# 1. Medidas de Tendencia Central
print(f"Media: {df['Edades'].mean():.2f}")
print(f"Mediana: {df['Edades'].median():.2f}")
print(f"Moda: {df['Edades'].mode()[0]}")
```

```python
# 2. Medidas de Dispersión
print(f"Varianza (ddof=1): {df['Edades'].var(ddof=1):.2f}")
print(f"Desviación estándar: {df['Edades'].std():.2f}")
print(f"Rango Intercuartílico (IQR): {stats.iqr(df['Edades']):.2f}")
print(f"Coeficiente de Variación: {stats.variation(df['Edades'])*100:.2f}%")
```

```python
# 3. Medidas de Forma
print(f"Asimetría (Skewness): {df['Edades'].skew():.2f}")
print(f"Curtosis (Kurtosis): {df['Edades'].kurt():.2f}")
```

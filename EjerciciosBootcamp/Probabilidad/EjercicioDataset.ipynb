```python
import numpy as np
```

```python
# Ejercicio 10: Dataset de clientes
# Creación de la población (1 = compra, 0 = no compra)
clientes = np.array([1] * 65 + [0] * 35)
```

```python
# 1 y 2. Cálculo de probabilidades teóricas
prob_compra = np.mean(clientes)
prob_no_compra = 1 - prob_compra

print("=== CÁLCULO TEÓRICO ===")
print(f"Probabilidad de compra: {prob_compra:.2f} ({prob_compra*100:.0f}%)")
print(f"Probabilidad de no compra: {prob_no_compra:.2f} ({prob_no_compra*100:.0f}%)")
print(f"Suma de probabilidades: {prob_compra + prob_no_compra:.2f}\n")
```

```python
# Reto extra: Simulación Montecarlo (10.000 clientes)
np.random.seed(42)  # Semilla para reproducibilidad
muestra = np.random.choice(clientes, size=10000)
prob_experimental = np.mean(muestra)

print("=== RETO EXTRA: SIMULACIÓN DE 10.000 CLIENTES ===")
print(f"Probabilidad experimental de compra: {prob_experimental:.4f} ({prob_experimental*100:.2f}%)")
```

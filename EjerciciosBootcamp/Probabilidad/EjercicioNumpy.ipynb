```python
import numpy as np
```

```python
# 1. Creación de la matriz 7x3 (7 días, 3 horas)
np.random.seed(42) #para resultados reproducibles
temperaturas = np.random.uniform(12.0, 32.0, size=(7,3))
```

```python
# 2. Inspección de forma y tipo
print("Matriz de temperaturas:\n", temperaturas)
print("Dimensione (shape):", temperaturas.shape)
print("Tipo de datos (dtype):", temperaturas.dtype)
```

```python
# Miércoles (fila índice 2) por la Tarde (columna índice 1)
temp_miercoles_tarde = temperaturas[2, 1]
print("Temperatura Miércoles por la tarde:", temp_miercoles_tarde)

# Noches (todas las filas, última columna índice 2)
temp_noches = temperaturas[:, 2]
print("Temperaturas por la Noche:", temp_noches)
```

```python
# Punto 4: Operaciones vectorizadas
# Conversión a Fahrenheit: F = C * 1.8 + 32
temperaturas_f = temperaturas * 1.8 + 32
print("Temperaturas en Fahrenheit:\n", temperaturas_f)

# Temperatura media de toda la semana
media_semanal = temperaturas.mean()
print("\nTemperatura media semanal (°C):", media_semanal)
```

```python
### Punto 5: Máscaras booleanas (Filtrado)
# Crear máscara booleana para valores > 25°C
mayores_25 = temperaturas[temperaturas > 25]
print("Temperaturas superiores a 25°C:\n", mayores_25)
```

```python
# Ejercicio 6: Distribución Normal y Regla Empírica
# 1. Generar la muestra
media, std = 170, 10
datos = np.random.normal(loc=media, scale=std, size=10000)

print(f"Media muestral: {np.mean(datos):.2f}")
print(f"Desviación estándar muestral: {np.std(datos):.2f}")
```

```python
# 2. Filtrar datos dentro de 1 desviación estándar
dentro_1_std = (datos >= media - std) & (datos <= media + std)
porcentaje = (np.sum(dentro_1_std) / len(datos)) * 100

print(f"Porcentaje dentro de 1 std: {porcentaje:.2f}%") # Debería dar cercano al 68%
```

```python
# Ejercicio 7: Teorema del Límite Central (Simulación)
# Muestras aleatorias de un dado uniformemente distribuido (1 a 6)
num_muestras = 1000
tamano_muestra = 30
```

```python
# Ejercicio 8: Probabilidad Condicional (Matriz de Confusión)
np.random.seed(42)
n = 1000

# 1 = Enfermo, 0 = Sano (5% enfermos)
enfermos = np.random.choice([1, 0], size=n, p=[0.05, 0.95])
```

```python
# Generar resultados del test
test = np.zeros(n)
for i in range(n):
    if enfermos[i] == 1:
        test[i] = np.random.choice([1, 0], p=[0.95, 0.05])
    else:
        test[i] = np.random.choice([1, 0], p=[0.10, 0.90])
```

```python
# P(Enfermo | Test Positivo)
positivos = (test == 1)
enfermos_y_positivos = (enfermos == 1) & (test == 1)

prob_condicional = np.sum(enfermos_y_positivos) / np.sum(positivos)
print(f"Probabilidad de estar enfermo dado test positivo: {prob_condicional * 100:.2f}%")
```

```python
# Ejercicio 9: Distribución Binomial (Control de Calidad)
# np.random.binomial(n_ensayos, prob_éxito, size)
defectuosas_por_lote = np.random.binomial(n=500, p=0.02, size=10000)

# Probabilidad de más de 15 defectuosas
lotes_criticos = np.sum(defectuosas_por_lote > 15)
probabilidad = lotes_criticos / len(defectuosas_por_lote)

print(f"Probabilidad de tener más de 15 defectuosas: {probabilidad * 100:.2f}%")
```

```python
# Ejercicio 10: Distribución de Poisson (Llegadas a un servidor)
# np.random.poisson(lam, size)
peticiones = np.random.poisson(lam=8, size=10000)

# 1. Exactamente 10 peticiones
prob_10 = np.mean(peticiones == 10)
print(f"Probabilidad de recibir exactamente 10 peticiones: {prob_10 * 100:.2f}%")

# 2. Más de 12 peticiones (sobrecarga)
prob_sobrecarga = np.mean(peticiones > 12)
print(f"Probabilidad de sobrecarga (>12 peticiones): {prob_sobrecarga * 100:.2f}%")
```

```python
# Ejercicio 11: Estimación de pi mediante el Método de Monte Carlo
n_puntos = 100000

# Puntos aleatorios entre -1 y 1
x = np.random.uniform(-1, 1, n_puntos)
y = np.random.uniform(-1, 1, n_puntos)

# Distancia al origen al cuadrado (x^2 + y^2)
dentro_circulo = (x**2 + y**2) <= 1

# Estimación de Pi: 4 * (puntos dentro / puntos totales)
pi_estimado = 4 * np.sum(dentro_circulo) / n_puntos

print(f"Valor estimado de Pi: {pi_estimado:.5f}")
print(f"Error absoluto: {abs(np.pi - pi_estimado):.5f}")
```

```python
# Ejercicio 12: Correlación y Covarianza
n = 1000
x = np.random.normal(50, 10, n)
ruido = np.random.normal(0, 5, n)

# Variable Y dependiente de X con cierto ruido
y = 2 * x + ruido

# 1. Matriz de Covarianza
matriz_cov = np.cov(x, y)
print("Matriz de Covarianza:\n", matriz_cov)

# 2. Matriz de Correlación
matriz_corr = np.corrcoef(x, y)
print("\nCoeficiente de correlación r(X, Y):", f"{matriz_corr[0, 1]:.4f}")
```

```python
# Ejercicio 13: Intervalos de Confianza con Bootstrap
muestra_original = np.array([12, 18, 25, 10, 15, 30, 22, 14, 19, 28])
n_bootstrap = 5000

# Generar 5000 muestras aleatorias con reemplazo del mismo tamaño
medias_bootstrap = np.zeros(n_bootstrap)

for i in range(n_bootstrap):
    resample = np.random.choice(muestra_original, size=len(muestra_original), replace=True)
    medias_bootstrap[i] = np.mean(resample)

# Percentiles 2.5 y 97.5 para el intervalo del 95%
ic_inferior = np.percentile(medias_bootstrap, 2.5)
ic_superior = np.percentile(medias_bootstrap, 97.5)

print(f"Media muestral original: {np.mean(muestra_original):.2f} min")
print(f"Intervalo de Confianza del 95%: [{ic_inferior:.2f}, {ic_superior:.2f}] min")
```

```python
# --- Ejercicio 4: Probabilidad Clásica (Dos dados) ---
dado1 = np.random.randint(1, 7, size=10000)
dado2 = np.random.randint(1, 7, size=10000)
suma = dado1 + dado2
prob_suma_7 = np.sum(suma == 7) / len(suma)
print(f"E4 - Probabilidad de sumar 7: {prob_suma_7 * 100:.2f}% (Teórica: 16.67%)")
```

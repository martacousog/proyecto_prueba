```python
# Ejercicio 1: Lanzamiento de Moneda
import numpy as np

# Lanzamiento de una moneda (0: Cara, 1: Cruz) 1000 veces
moneda = np.random.choice([0, 1], size=1000, p=[0.5, 0.5])
caras = np.sum(moneda == 0)
cruces = np.sum(moneda == 1)

print(f"Caras: {caras} ({caras/10:.1f}%), Cruces: {cruces} ({cruces/10:.1f}%)")
```

```python
# Ejercicio 2: Estadísticos Básicos
datos_ventas = np.random.exponential(scale=50, size=500)

print(f"Media: {np.mean(datos_ventas):.2f}")
print(f"Mediana: {np.median(datos_ventas):.2f}")
print(f"Percentil 25: {np.percentile(datos_ventas, 25):.2f}")
print(f"Percentil 75: {np.percentile(datos_ventas, 75):.2f}")
```

```python
# Ejercicio 3: Varianza y Desviación Estándar
muestra_a = np.random.normal(loc=100, scale=5, size=200)
muestra_b = np.random.normal(loc=100, scale=20, size=200)

print(f"Desviación Estándar Muestra A: {np.std(muestra_a):.2f}")
print(f"Desviación Estándar Muestra B: {np.std(muestra_b):.2f}")
```

```python
# Ejercicio 4: Probabilidad Clásica (Dos Dados)
dado1 = np.random.randint(1, 7, size=10000)
dado2 = np.random.randint(1, 7, size=10000)
suma = dado1 + dado2

prob_suma_7 = np.sum(suma == 7) / len(suma)
print(f"Probabilidad de sumar 7: {prob_suma_7 * 100:.2f}%")
```

```python
# Ejercicio 5: Normalización Z-Score
precios = np.array([120, 250, 90, 450, 310, 180, 290])
z_scores = (precios - np.mean(precios)) / np.std(precios)

print(f"Z-Scores normalizados:\n{z_scores.round(2)}")
```

```python
# Ejercicio 6: Distribución Normal
media, std = 170, 10
datos_normales = np.random.normal(loc=media, scale=std, size=10000)
dentro_1_std = (datos_normales >= media - std) & (datos_normales <= media + std)

print(f"Porcentaje dentro de 1 std: {np.mean(dentro_1_std) * 100:.2f}%")
```

```python
# Ejercicio 1. Moneda
import numpy as np

monedas = np.random.choice([0, 1], 1000)
frecuencia_caras = monedas.mean()

print(f"Frecuencia relativa de caras: {frecuencia_caras}")
```

```python
# Ejercicio 2. Dado
dados = np.random.randint(1, 7, 10000)

# Frecuencia de cada valor (del 1 al 6)
valores, frecuencias = np.unique(dados, return_counts=True)

for val, freq in zip(valores, frecuencias):
    print(f"Valor {val}: {freq} veces ({freq/10000:.2%})")
```

```python
# Ejercicio 3. Dos dados
d1 = np.random.randint(1, 7, 100000)
d2 = np.random.randint(1, 7, 100000)

prob_suma_7 = ((d1 + d2) == 7).mean()
print(f"Probabilidad de suma 7: {prob_suma_7:.4f}")
```

```python
# Ejercicio 4. Bolsa de bolas
bolsa = np.array(["R"] * 4 + ["A"] * 3 + ["V"] * 3)

# Simulamos 10.000 extracciones aleatorias con reemplazo
extracciones = np.random.choice(bolsa, 10000)

prob_roja = (extracciones == "R").mean()
prob_azul = (extracciones == "A").mean()
prob_verde = (extracciones == "V").mean()

print(f"P(Roja): {prob_roja:.2%}, P(Azul): {prob_azul:.2%}, P(Verde): {prob_verde:.2%}")
```

```python
# Ejercicio 5. Dataset clientes
clientes = np.array([1] * 65 + [0] * 35)

# Simulamos un muestreo aleatorio de 1.000 clientes
muestra = np.random.choice(clientes, 1000)
prob_compra = muestra.mean()

print(f"Probabilidad estimada de compra: {prob_compra:.2%}")
```

```python
# Ejercicio 6. Spam
correos = np.array([1] * 80 + [0] * 20)

# Simulamos recibir 5.000 correos
simulacion = np.random.choice(correos, 5000)
prob_spam = simulacion.mean()

print(f"Probabilidad de que un correo sea spam: {prob_spam:.2%}")
```

```python
# Ejercicio 7. Lotería
# Compramos 10 décimos distintos
mis_numeros = np.random.choice(100000, 10, replace=False)

# Simulación de 10.000 sorteos
sorteos = np.random.randint(0, 100000, 10000)

# Comprobamos cuántas veces alguno de nuestros números salió premiado
ganados = np.isin(sorteos, mis_numeros).sum()
prob_ganar = ganados / 10000

print(f"Probabilidad de ganar al menos un sorteo: {prob_ganar:.4%}")
```

```python
# Ejercicio 8. Ley de los grandes números
for n in [100, 1000, 10000]:
    lanzamientos = np.random.choice([0, 1], n)
    media = lanzamientos.mean()
    print(f"Con N={n:5d} -> Frecuencia de caras: {media:.4f} (Diferencia con 0.5: {abs(0.5 - media):.4f})")
    
```

```python
# Ejercicio 9. Histograma
import matplotlib.pyplot as plt

dados = np.random.randint(1, 7, 10000)

plt.hist(dados, bins=np.arange(1, 8) - 0.5, rwidth=0.8, color='skyblue', edgecolor='black')
plt.title("Frecuencia de lanzamientos de un dado (10,000 tiradas)")
plt.xlabel("Valor del dado")
plt.ylabel("Frecuencia")
plt.xticks(range(1, 7))
plt.show()
```

```python
# Ejercicio 10. Reto Final
# Probabilidad teórica de ganar la Lotería de Navidad con 1 décimo
prob_teorica = 1 / 100000  # 0.00001 (0.001%)

# Simulamos 50.000 sorteos y elegimos un número fijo (ej. el 12345)
numero_comprado = 12345
sorteos = np.random.randint(0, 100000, 50000)

# Calculamos probabilidad experimental
aciertos = (sorteos == numero_comprado).sum()
prob_experimental = aciertos / 50000

print(f"Probabilidad Teórica:    {prob_teorica:.5f}")
print(f"Probabilidad Experimental: {prob_experimental:.5f} (Ganaste {aciertos} veces)")
```

```python
import numpy as np
```

```python
# 1. Creación de la matriz 7x3 (7 días, 3 horas)
np.random.seed(42) #para resultados reproducibles
temperaturas = np.random.uniform(12.0, 32.0, size=(7,3))
```

```python
# 2. Inspección de forma y tipo
print("Matriz de temperaturas:\n", temperaturas)
print("Dimensione (shape):", temperaturas.shape)
print("Tipo de datos (dtype):", temperaturas.dtype)
```

```python
# Miércoles (fila índice 2) por la Tarde (columna índice 1)
temp_miercoles_tarde = temperaturas[2, 1]
print("Temperatura Miércoles por la tarde:", temp_miercoles_tarde)

# Noches (todas las filas, última columna índice 2)
temp_noches = temperaturas[:, 2]
print("Temperaturas por la Noche:", temp_noches)
```

```python
# Punto 4: Operaciones vectorizadas
# Conversión a Fahrenheit: F = C * 1.8 + 32
temperaturas_f = temperaturas * 1.8 + 32
print("Temperaturas en Fahrenheit:\n", temperaturas_f)

# Temperatura media de toda la semana
media_semanal = temperaturas.mean()
print("\nTemperatura media semanal (°C):", media_semanal)
```

```python
### Punto 5: Máscaras booleanas (Filtrado)
# Crear máscara booleana para valores > 25°C
mayores_25 = temperaturas[temperaturas > 25]
print("Temperaturas superiores a 25°C:\n", mayores_25)
```

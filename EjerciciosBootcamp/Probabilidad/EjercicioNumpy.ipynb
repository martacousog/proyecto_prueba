```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Fijamos la semilla para obtener siempre los mismos resultados
np.random.seed(42)
```

```python
# Ejercicio 1. Moneda (1.000 lanzamientos)
monedas = np.random.choice([0, 1], size=1000)
print("=== EJERCICIO 1: MONEDA ===")
print("Frecuencia relativa de caras (1):", monedas.mean())
```

```python
# Ejercicio 2. Dado (10.000 tiradas)
dados = np.random.randint(1, 7, size=10000)
valores, frecuencias = np.unique(dados, return_counts=True)
print("=== EJERCICIO 2: DADO ===")
for val, freq in zip(valores, frecuencias):
    print(f"Valor {val}: {freq} veces ({freq/10000:.4f})")
```

```python
# Ejercicio 3. Dos dados (Suma igual a 7)
d1 = np.random.randint(1, 7, size=100000)
d2 = np.random.randint(1, 7, size=100000)
prob_suma_7 = ((d1 + d2) == 7).mean()
print("=== EJERCICIO 3: DOS DADOS ===")
print("Probabilidad experimental de suma 7:", prob_suma_7)
```

```python
# Ejercicio 4. Bolsa de bolas (Extracción aleatoria)
bolsa = np.array(["R"] * 4 + ["A"] * 3 + ["V"] * 3)
extracciones = np.random.choice(bolsa, size=10000)
print("=== EJERCICIO 4: BOLSA DE BOLAS ===")
print("P(Roja):", (extracciones == "R").mean())
print("P(Azul):", (extracciones == "A").mean())
print("P(Verde):", (extracciones == "V").mean())
```

```python
# Ejercicio 6. Spam (Probabilidad de Spam)
correos = np.array([1] * 80 + [0] * 20)
muestras_correos = np.random.choice(correos, size=10000)
print("=== EJERCICIO 6: SPAM ===")
print("Probabilidad estimada de Spam:", muestras_correos.mean())
```

```python
# Ejercicio 7. Lotería (100.000 números, 10 décimos)
mis_numeros = np.random.choice(100000, size=10, replace=False)
sorteo = np.random.randint(0, 100000)
ganado = np.isin(sorteo, mis_numeros)
print("=== EJERCICIO 7: LOTERÍA ===")
print("Mis números:", mis_numeros)
print("Número premiado:", sorteo)
print("¿Premio obtenido?:", ganado)
```

```python
# Ejercicio 8. Ley de los grandes números (Simulaciones)
print("=== EJERCICIO 8: LEY DE LOS GRANDES NÚMEROS ===")
for n in [100, 1000, 10000, 100000]:
    simulacion = np.random.choice([0, 1], size=n)
    print(f"N = {n:6d} | Frecuencia estimada de cara: {simulacion.mean():.4f}")
```

```python
# Ejercicio 9. Histograma (Representación gráfica)
plt.figure(figsize=(7, 4))
plt.hist(dados, bins=np.arange(1, 8) - 0.5, rwidth=0.8, color='skyblue', edgecolor='black')
plt.title("Ejercicio 9: Frecuencia de Tiradas de Dado (10.000 simulaciones)")
plt.xlabel("Cara del dado")
plt.ylabel("Frecuencia absoluta")
plt.xticks(range(1, 7))
```

```python
# Ejercicio 10. Reto Final (50.000 sorteos de Navidad)
sorteos = np.random.randint(0, 100000, size=50000)
victorias = np.sum(np.isin(sorteos, mis_numeros))
prob_exp_navidad = victorias / 50000
prob_teor_navidad = 10 / 100000

print("=== EJERCICIO 10: RETO FINAL NAVIDAD ===")
print(f"Probabilidad Teórica: {prob_teor_navidad:.5f} ({prob_teor_navidad*100:.3f}%)")
print(f"Probabilidad Experimental (50k sorteos): {prob_exp_navidad:.5f} ({prob_exp_navidad*100:.3f}%)")
```

```python
# Bloque 1: Conceptos básicos

# Ejercicio 1: Moneda
p_cara = 1 / 2
p_cruz = 1 / 2
suma_p = p_cara + p_cruz
print(f"E1 - P(Cara): {p_cara}, P(Cruz): {p_cruz}, Suma: {suma_p}")
```

```python
# Ejercicio 2: Dado (6 caras)
p_cuatro = 1 / 6
p_par = 3 / 6
p_mayor_4 = 2 / 6
p_menor_igual_5 = 5 / 6
print(f"E2 - P(4): {p_cuatro:.3f}, P(Par): {p_par}, P(>4): {p_mayor_4:.3f}, P(<=5): {p_menor_igual_5:.3f}")
```

```python
# Ejercicio 3: Baraja española (40 cartas)
p_as = 4 / 40
p_rey = 4 / 40
p_copas = 10 / 40
p_figura = 12 / 40
print(f"E3 - P(As): {p_as}, P(Rey): {p_rey}, P(Copas): {p_copas}, P(Figura): {p_figura}")
```

```python
# Bloque 2: Probabilidad simple

# Ejercicio 4: Bolsa de bolas
total_bolas = 4 + 3 + 3
p_roja = 4 / total_bolas
p_azul = 3 / total_bolas
p_verde = 3 / total_bolas
print(f"E4 - P(Roja): {p_roja:.2f}, P(Azul): {p_azul:.2f}, P(Verde): {p_verde:.2f}")
```

```python
# Ejercicio 5: Alumnos Bootcamp
total_alumnos = 12 + 8
p_ds = 12 / total_alumnos
p_ia = 8 / total_alumnos
print(f"E5 - P(Data Science): {p_ds:.2f}, P(IA): {p_ia:.2f}\n")
```

```python
# Bloque 3: Eventos complementarios

# Ejercicio 6: Contraseñas (dígitos 0 al 9)
p_par = 5 / 10
p_no_par = 1 - p_par
print(f"E6 - P(Par): {p_par:.2f}, P(No Par): {p_no_par:.2f}")
```

```python
# Ejercicio 7: Lotería simplificada
p_ganar = 1 / 1000
p_perder = 1 - p_ganar
print(f"E7 - P(Ganar): {p_ganar:.3f}, P(Perder): {p_perder:.3f}")
```

```python
# Bloque 4: Probabilidad conjunta

# Ejercicio 8: Dos monedas
espacio_monedas = [('C', 'C'), ('C', 'X'), ('X', 'C'), ('X', 'X')]
p_dos_caras = 1 / len(espacio_monedas)
p_al_menos_una_cara = 3 / len(espacio_monedas)
print(f"E8 - P(Dos caras): {p_dos_caras:.2f}, P(Al menos una cara): {p_al_menos_una_cara:.2f}")
```

```python
# Ejercicio 9: Dos dados (36 resultados)
p_suma_7 = 6 / 36
p_suma_mayor_10 = 3 / 36
print(f"E9 - P(Suma = 7): {p_suma_7:.3f}, P(Suma > 10): {p_suma_mayor_10:.3f}\n")
```

```python
# Bloque 5: Introducción a Data Science

# Ejercicio 10: Dataset de clientes
total_clientes = 65 + 35
p_compra = 65 / total_clientes
p_no_compra = 35 / total_clientes
print(f"E10 - P(Compra): {p_compra:.2f}, P(No compra): {p_no_compra:.2f}")
```

```python
# Ejercicio 11: Clasificación binaria
total_correos = 80 + 20
p_spam = 80 / total_correos
p_no_spam = 20 / total_correos
print(f"E11 - P(Spam): {p_spam:.2f}, P(No Spam): {p_no_spam:.2f}")
```
